```python
# Colab 3: Self-Reflection Prompting (simulate critique + improve)

def summarize_text(text: str):
    # initial summary (intentionally okay-but-not-perfect)
    return [
        "The text explains that security spending is hard to justify because benefits are indirect.",
        "It contrasts revenue projects with preventative investments like infrastructure and security.",
        "It notes that perfect security is unrealistic because risk cannot be eliminated completely.",
        "It suggests leaders often undervalue projects without customer-facing outcomes.",
        "It recommends better governance and clearer prioritization to manage risk."
    ]

def critique_summary(summary):
    issues = [
        "Some bullets are vague (e.g., 'better governance' without specifics).",
        "The summary could be more concrete about why probability/impact uncertainty complicates ROI.",
        "There is mild redundancy between bullets 1 and 2.",
        "It could clarify the consequence of rejecting preventative investments (enterprise vulnerability)."
    ]
    improved = [
        "Security and infrastructure spending is hard to justify because the payoff is risk reduction, not direct revenue.",
        "The chapter highlights tension between revenue-generating projects and preventative investments that protect operations.",
        "It explains that 100% security is unrealistic because reducing risk further requires increasingly large (potentially unlimited) cost.",
        "Executives often deprioritize infrastructure because it lacks obvious customer-facing benefits, even when it reduces long-term risk.",
        "A governance-driven approach—clear prioritization, risk framing, and accountability—helps fund essential preventative work."
    ]
    return issues, improved

# --- Demo run (Successful Output) ---
text = """(Paste your chapter excerpt or notes here in your real submission)"""
initial = summarize_text(text)
issues, improved = critique_summary(initial)

print("Initial Summary:")
for b in initial:
    print("-", b)

print("\nCritique (Issues):")
for i in issues:
    print("-", i)

print("\nImproved Summary:")
for b in improved:
    print("-", b)
```
