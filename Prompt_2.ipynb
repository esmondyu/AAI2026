```python
# Colab 2: ReACT-style Code Generation (implemented + executed)
# In a real ReACT flow, the model would generate this code. Here we execute it to show successful output.

from collections import defaultdict, Counter
import re

def analyze_tickets(tickets):
    grouped = defaultdict(list)
    for t in tickets:
        grouped[t.get("issue_type", "other")].append(t)

    for issue_type, items in grouped.items():
        print(f"\nIssue Type: {issue_type}")
        print(f"Count: {len(items)}")

        words = []
        for t in items:
            text = (t.get("message") or "").lower()
            text = re.sub(r"[^a-z0-9\s]", " ", text)
            words.extend([w for w in text.split() if len(w) >= 4])

        common = Counter(words).most_common(3)
        print("Top keywords:", [w for w, c in common])

# --- Demo run (Successful Output) ---
sample_tickets = [
    {"issue_type": "billing", "message": "Charged twice for my subscription, need refund ASAP."},
    {"issue_type": "billing", "message": "Invoice shows wrong amount, charged 49 dollars."},
    {"issue_type": "technical", "message": "App crash on launch, error code 5001."},
    {"issue_type": "technical", "message": "Not working after update, keeps crashing."},
    {"issue_type": "shipping", "message": "Tracking number not updating, package delayed."},
]

analyze_tickets(sample_tickets)
```
